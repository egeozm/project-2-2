```python
# Extract country codes from column names
country_codes = set()
for col in df.columns:
    if col != 'cet_cest_timestamp':
        country_code = col.split('_')[0]
        country_codes.add(country_code)

print(f"Countries in the dataset: {sorted(country_codes)}")

# For demonstration, select a few important countries to analyze (including ones with weather data)
countries_to_analyze = ['AT', 'ES', 'FR', 'IT', 'NO']

# Create a function to plot time series for a specific country
def plot_country_data(country_code, n_months=3):
    # Get columns for this country
    country_cols = [col for col in df.columns if col.startswith(f"{country_code}_")]
    
    if len(country_cols) == 0:
        print(f"No data columns found for {country_code}")
        return
        
    # Use a sample of data (first n months) to make visualization clearer
    sample_df = df.iloc[:n_months*7*24]  # Approx n months of hourly data
    
    # Create matplotlib figure with subplots
    fig, axes = plt.subplots(len(country_cols), 1, figsize=(12, 3*len(country_cols)), sharex=True)
    fig.suptitle(f"{country_code} Energy Data Time Series", fontsize=16)
    
    # If there's only one column, axes is not a list, so we convert it
    if len(country_cols) == 1:
        axes = [axes]
    
    # Plot each column as a separate subplot
    for i, col in enumerate(country_cols):
        axes[i].plot(sample_df.index, sample_df[col])
        axes[i].set_title(col)
        axes[i].grid(True)
    
    plt.tight_layout(rect=[0, 0, 1, 0.97])  # Adjust to make room for the super title
    plt.show()
    
    # Calculate and display summary statistics
    stats_df = pd.DataFrame({
        'Column': country_cols,
        'Missing (%)': [sample_df[col].isna().mean() * 100 for col in country_cols],
        'Min': [sample_df[col].min() for col in country_cols],
        'Max': [sample_df[col].max() for col in country_cols],
        'Mean': [sample_df[col].mean() for col in country_cols],
        'Std': [sample_df[col].std() for col in country_cols]
    })
    
    display(stats_df)
```

## Step 3: Visualize data gaps across time

## Step 5: Identify Anomalies in the Time Series Data

Let's look for anomalies and outliers in the energy data using statistical methods.

```python
# Function to detect and visualize anomalies using Z-score method
def detect_anomalies(country, data_type):
    # Find relevant columns
    cols = [col for col in df.columns if col.startswith(f"{country}_") and data_type in col]
    
    if not cols:
        print(f"No {data_type} data found for {country}")
        return
    
    # For each column, detect anomalies
    for col in cols:
        # Get data without NaN values
        data = df[col].dropna()
        
        if len(data) == 0:
            print(f"No valid data in {col}")
            continue
        
        # Calculate Z-scores
        z_scores = (data - data.mean()) / data.std()
        
        # Define anomalies as data points with Z-score > 3 or < -3
        anomalies = data[abs(z_scores) > 3]
        
        # Visualization
        if len(anomalies) > 0:
            plt.figure(figsize=(12, 6))
            
            # Original data (show a subset for clarity)
            sample_data = data.iloc[:8760]  # ~1 year of hourly data
            plt.plot(sample_data.index, sample_data, 'b-', alpha=0.5, label=col)
            
            # Highlight anomalies that fall within the sample period
            sample_anomalies = anomalies[anomalies.index.isin(sample_data.index)]
            if len(sample_anomalies) > 0:
                plt.scatter(sample_anomalies.index, sample_anomalies, color='red', s=50, label='Anomalies')
            
            plt.title(f'Anomalies in {col}')
            plt.legend()
            plt.grid(True)
            plt.tight_layout()
            plt.show()
            
            # Print anomaly statistics
            print(f"Found {len(anomalies)} anomalies in {col}:")
            display(anomalies.describe())
        else:
            print(f"No significant anomalies found in {col}")

# Detect anomalies in load data for various countries
detect_anomalies('AT', 'load_actual')
```

## Step 6: Implement Strategies for Handling Missing Data

Based on our analysis of missing data patterns by country, year, and data type, we'll implement several strategies to handle these gaps:

1. **Country Selection**: Filter to focus only on countries with high data completeness
2. **Time Period Selection**: Exclude periods with excessive missing data (e.g., 2020 COVID period)
3. **Feature Selection**: Focus on metrics with better coverage
4. **Imputation Techniques**: Apply appropriate techniques for filling gaps in time series data

```python
# Strategy 1: Select countries with high data completeness
# Filter for countries with less than 5% missing data
good_data_countries = [country for country, (pct, _, _, _) 
                       in country_completeness.items() if pct < 5.0]

print(f"Countries with excellent data quality (less than 5% missing):\n{sorted(good_data_countries)}")

# Strategy 2: Select optimal time periods
# Exclude 2020 (COVID period with 17.27% missing data)
optimal_time_range = ('2015-01-01', '2019-12-31')
df_filtered_time = df.loc[optimal_time_range[0]:optimal_time_range[1]]
print(f"\nSelected time range: {optimal_time_range[0]} to {optimal_time_range[1]}")
print(f"Number of data points before filtering: {len(df)}")
print(f"Number of data points after time filtering: {len(df_filtered_time)}")

# Calculate percentage of missing data in the filtered timeframe
missing_before = df.isna().sum().sum() / df.size * 100
missing_after_time_filter = df_filtered_time.isna().sum().sum() / df_filtered_time.size * 100
print(f"Missing data before filtering: {missing_before:.2f}%")
print(f"Missing data after time filtering: {missing_after_time_filter:.2f}%")
```

```python
# Strategy 3: Create a clean dataset with selected countries and features
# Focus on most reliable countries and most complete features

# For each selected country, identify the most complete metrics
reliable_features = []
for country in good_data_countries:
    # Get all columns for this country
    country_cols = [col for col in df.columns if col.startswith(f"{country}_")]
    
    # Get missing percentage for each column
    col_missing = {col: df_filtered_time[col].isna().mean() * 100 for col in country_cols}
    
    # Keep features with less than 5% missing values
    good_cols = [col for col, missing_pct in col_missing.items() if missing_pct < 5.0]
    reliable_features.extend(good_cols)

# Create a clean dataset with reliable features
df_clean = df_filtered_time[reliable_features]

print(f"\nSelected {len(reliable_features)} reliable features from {len(good_data_countries)} high-quality countries")
print(f"Clean dataset shape: {df_clean.shape}")
print(f"Missing data in clean dataset: {df_clean.isna().sum().sum() / df_clean.size * 100:.2f}%")

# Display the first few columns to check what we have
print("\nSample of selected features:")
print(reliable_features[:10])
```

```python
# Strategy 4: Implement imputation strategies for filling remaining gaps

# Function to apply different imputation methods and compare results
def apply_imputation(df, columns=None):
    """Apply different imputation methods to time series data
    and compare results with visualizations"""
    if columns is None:
        # If no columns specified, choose first column as example
        columns = [df.columns[0]]
    
    # Create a copy of original data for comparison
    results = {}
    
    for col in columns:
        # Skip columns with no missing data
        if df[col].isna().sum() == 0:
            print(f"No missing data in {col}. Skipping...")
            continue
            
        print(f"\nProcessing column: {col}")
        print(f"Missing values: {df[col].isna().sum()} ({df[col].isna().mean()*100:.2f}%)")
        
        # Create copies with different imputation methods
        results[col] = {
            'original': df[col].copy(),
            'forward_fill': df[col].ffill(),
            'backward_fill': df[col].bfill(),
            'linear_interp': df[col].interpolate(method='linear'),
            'time_interp': df[col].interpolate(method='time')  # Good for time series
        }
        
        # Add rolling mean imputation
        # First create a copy with linear interpolation, then use rolling window
        interpolated = df[col].interpolate(method='linear')
        # For remaining NAs at the edges, use rolling mean
        window_size = 24  # 24 hours rolling window
        rolling_mean = interpolated.rolling(window=window_size, center=True, min_periods=1).mean()
        # Fill any remaining NAs with the rolling mean
        results[col]['rolling_mean'] = interpolated.fillna(rolling_mean)
        
        # Create visualization to compare methods
        # Find a span with missing values to visualize
        na_indices = df[col].isna()
        if na_indices.any():
            # Get a sample period around the missing values
            na_idx = na_indices.idxmax()
            # 5 days before and after the NA value
            sample_start = na_idx - pd.Timedelta(days=5)
            sample_end = na_idx + pd.Timedelta(days=5)
            
            # Plot the different imputation methods
            plt.figure(figsize=(15, 8))
            
            for method, data in results[col].items():
                # Show a sample period centered around NAs for clarity
                sample = data.loc[sample_start:sample_end]
                plt.plot(sample.index, sample, label=method, marker='o' if method == 'original' else None)
            
            plt.title(f'Comparison of Imputation Methods for {col}')
            plt.legend()
            plt.grid(True)
            plt.tight_layout()
            plt.show()
    
    return results

# Test imputation on a few example columns from ES (Spain, which has good overall data)
country_to_test = 'ES'
test_cols = [col for col in df_clean.columns if col.startswith(f"{country_to_test}_")][:2]  # Test first two columns
if test_cols:
    imputation_results = apply_imputation(df_clean, test_cols)
else:
    print(f"No columns found for {country_to_test} in the clean dataset")
```

```python
# Strategy 5: Create the final cleaned dataset with optimal imputation

# Based on the visualization results, choose the best imputation method
# Time-based interpolation is usually best for time series data

# Apply time-based interpolation to all columns in the clean dataset
df_final = df_clean.copy()

# Apply time-based interpolation to each column
for col in df_final.columns:
    na_count_before = df_final[col].isna().sum()
    if na_count_before > 0:
        # Try time-based interpolation first
        df_final[col] = df_final[col].interpolate(method='time')
        
        # For any remaining NAs (typically at edges), use forward fill then backward fill
        df_final[col] = df_final[col].ffill().bfill()
        
        na_count_after = df_final[col].isna().sum()
        print(f"Column {col}: Filled {na_count_before - na_count_after} missing values")

# Verify no missing values remain
remaining_nas = df_final.isna().sum().sum()
print(f"\nRemaining missing values in final dataset: {remaining_nas}")
print(f"Final clean dataset shape: {df_final.shape}")

# Save clean dataset to CSV for future use
clean_csv_path = os.path.join(os.path.dirname(os.path.dirname(os.path.abspath('__file__'))), 
                           'clean_energy_data.csv')

try:
    df_final.to_csv(clean_csv_path)
    print(f"\nSaved clean dataset to: {clean_csv_path}")
except Exception as e:
    print(f"Error saving clean dataset: {e}")
```

## Step 7: Verify Cleaned Data - Slovenia Case Study

Let's analyze a random week of data from Slovenia (SI) to verify the quality of our cleaned dataset. Slovenia is one of the countries with excellent data quality (less than 0.1% missing data in the original dataset).

```python
# Plot all available Slovenia data for a random week from the cleaned dataset
import random
from datetime import timedelta

# Check what columns we have for Slovenia in the cleaned dataset
si_columns = [col for col in df_final.columns if col.startswith('SI_')]
print(f"Available features for Slovenia in cleaned dataset: {len(si_columns)}")
print(si_columns)

# Select a random week (Sunday to Saturday) from the dataset
# Excluding the first and last 2 weeks to ensure we have complete data
start_date = df_final.index[14]  # Start from day 14 (after 2 weeks)
end_date = df_final.index[-14]  # End 2 weeks before the last day

# Create a list of possible week start dates (Sundays)
possible_starts = [date for date in df_final.index if start_date <= date <= end_date and date.dayofweek == 6]

# Select a random Sunday as the start of our week
random_sunday = random.choice(possible_starts)
week_start = random_sunday
week_end = week_start + timedelta(days=7)

print(f"\nSelected random week: {week_start.date()} to {week_end.date()}")

# Get data for this week
si_week_data = df_final.loc[week_start:week_end, si_columns]

# Create a figure with multiple subplots - one for each feature
plt.figure(figsize=(15, 4 * len(si_columns)))

# Plot each feature
for i, column in enumerate(si_columns, 1):
    plt.subplot(len(si_columns), 1, i)
    plt.plot(si_week_data.index, si_week_data[column], marker='o', linestyle='-', markersize=4)
    plt.title(f"{column}")
    plt.grid(True)
    
    # Add day of week labels
    days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
    for j in range(7):
        day_date = week_start + timedelta(days=j)
        plt.axvline(x=day_date, color='gray', linestyle='--', alpha=0.3)
        plt.text(day_date, plt.ylim()[1] * 0.95, days[day_date.dayofweek], 
                 ha='center', va='top', backgroundcolor='white')

plt.tight_layout()
plt.show()

# Generate summary statistics for the week
summary_stats = si_week_data.describe().T
summary_stats['column'] = summary_stats.index
summary_stats = summary_stats.reset_index(drop=True)
summary_stats = summary_stats[['column', 'mean', 'std', 'min', '25%', '50%', '75%', 'max']]
print("\nSummary statistics for the week:")
display(summary_stats)
```

```python
## YEARLY PLOT

# Get data for a full year
year_start = df_final.index[0]
year_end = year_start + pd.DateOffset(years=1)
si_year_data = df_final.loc[year_start:year_end, si_columns]
si_year_daily = si_year_data.resample('D').mean()

# Plot with better spacing and cleaner visuals
fig, axes = plt.subplots(len(si_columns), 1, figsize=(16, 3 * len(si_columns)), sharex=True)

for i, column in enumerate(si_columns):
    ax = axes[i] if len(si_columns) > 1 else axes
    ax.plot(si_year_daily.index, si_year_daily[column], color='steelblue', linewidth=1.5)
    ax.set_title(column.replace("SI_", "").replace("_", " ").title(), fontsize=12)
    ax.grid(True, which='major', linestyle='--', alpha=0.5)

# Formatting x-axis with fewer ticks
fig.autofmt_xdate()
plt.tight_layout()
plt.show()
```
